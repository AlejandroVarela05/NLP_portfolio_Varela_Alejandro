### Exercise 1.2: Heaps' Law Demonstration

The slides mention Heaps' Law: vocabulary size grows with √N.

Generate text of increasing length and observe vocabulary growth.

```python
# YOUR CODE HERE
import random

# Use a simple word list to simulate text
word_list = ['the', 'a', 'is', 'are', 'was', 'were', 'be', 'been',
             'cat', 'dog', 'bird', 'fish', 'tree', 'house', 'car',
             'run', 'walk', 'jump', 'eat', 'sleep', 'read', 'write',
             'big', 'small', 'fast', 'slow', 'red', 'blue', 'green']

for i in range(1, 6):
    text_length = 10 ** i
    text = ' '.join(random.choices(word_list, k=text_length))
    tokens = text.split()
    types = set(tokens)
    print(f"Text length: {text_length}, Vocabulary size: {len(types)}")
```

### Exercise 1.3: BPE Simulation

The slides explain Byte Pair Encoding (BPE).
Implement a simple BPE token learner that:
1. Starts with character vocabulary
2. Finds most frequent adjacent pair
3. Merges them into a new token
4. Repeats k times

```python
# YOUR CODE HERE
from os import times

corpus = "low lower newest widest"

def simple_bpe(corpus, num_merges): 
    # Step 1: Initialize vocabulary with characters
    tokens = corpus.split()
    vocab = set(' '.join(tokens))  # Start with individual characters
    
    for _ in range(num_merges):
        # Step 2: Count frequency of adjacent pairs
        pair_freq = {}
        for token in tokens:
            for i in range(len(token) - 1):
                pair = (token[i], token[i + 1])
                pair_freq[pair] = pair_freq.get(pair, 0) + 1
        
        # Step 3: Find the most frequent pair
        if not pair_freq:
            break
        most_frequent_pair = max(pair_freq, key=pair_freq.get)
        
        # Step 4: Merge the most frequent pair into a new token
        new_token = ''.join(most_frequent_pair)
        vocab.add(new_token)
        
        # Update tokens by replacing the most frequent pair with the new token
        new_tokens = []
        for token in tokens:
            new_tokenized = token.replace(''.join(most_frequent_pair), new_token)
            new_tokens.append(new_tokenized)
        tokens = new_tokens
    
    return vocab
corpus = "low lower newest widest"
num_merges = 5
vocab = simple_bpe(corpus, num_merges)
print("Learned BPE Vocabulary:", vocab)
```

### Exercise 2.1: Lookahead Assertions

The slides introduce lookahead: (?=pattern) and (?!pattern)

Write patterns to:
1. Find words followed by a comma (without capturing comma)
2. Find first word of line only if it doesn't start with 'T'

```python
# YOUR CODE HERE
import re

text = "The quick, brown fox jumps over the lazy dog."
text2 = "Others, however, are not so lucky."
# 1. Find words followed by a comma (without capturing comma)
pattern1 = r'\b\w+\b(?=,)'
print(re.findall(pattern1, text))
# 2. Find first word of line only if it doesn't start with 'T'
pattern2 = r'^(?!T)\b\w+\b'
print(re.findall(pattern2, text))
print(re.findall(pattern2, text2))
```

### Exercise 2.2: Non-capturing Groups

The slides explain (?:...) for grouping without capturing.

Write a pattern that matches "some cats" or "a few cats" 
but only captures "cats" (not "some" or "a few").

```python
# YOUR CODE HERE
import re


texts = [
    "some cats like fish",
    "a few cats play outside", 
    "some dogs bark"
]

pattern = r'(?:some|a few)\s+(cats)'
print(re.findall(pattern, ' '.join(texts)))
```

### Exercise 3.1: Identifying Morphemes

The slides define morphemes as minimal meaning-bearing units.

Write code to identify potential morphemes by finding:
1. Common suffixes (-ed, -ing, -s, -ly, -ful)
2. Common prefixes (un-, re-, pre-, dis-)

```python
# YOUR CODE HERE
from sys import prefix

words = ["working", "unhappy", "carefully", "reworked", "glasses", "preprocessing"]

# Write code to identify potential morphemes by finding:
# 1. Common suffixes (-ed, -ing, -s, -ly, -ful)
suffixes = ['ed', 'ing', 's', 'ly', 'ful']
for word in words:
    for suffix in suffixes:
        if word.endswith(suffix):
            print(f"For the 1. -> Word: {word}, Suffix: {suffix}, Stem: {word[:-len(suffix)]}")
# 2. Common prefixes (un-, re-, pre-, dis-)
prefixes = ['un', 're', 'pre', 'dis']
for word in words:
    for prefix in prefixes:
        if word.startswith(prefix):
            print(f"For the 2. -> Word: {word}, Prefix: {prefix}, Stem: {word[len(prefix):]}")
```
