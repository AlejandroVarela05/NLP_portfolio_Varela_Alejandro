```python
# 3. Check if string contains only alphabets
def exercise_1_3(s):
    # Return True if all characters in the string are letters, False otherwise
    return s.isalpha()

# Test the function with 'hello' (should be True)
print(f"3. 'hello' is alpha: {exercise_1_3('hello')}")
# Test the function with 'hello123' (should be False)
print(f"   'hello123' is alpha: {exercise_1_3('hello123')}")
```

```python
# 4. Reverse a string without slicing
def exercise_1_4(s):
    # Create an empty string to store the result
    result = ""
    # Loop through each character in the string
    for char in s:
        # Add each character to the front of the result string
        result = char + result
    # Return the reversed string
    return result

# Test the function with 'hello'
print(f"4. Reversed 'hello': {exercise_1_4('hello')}")
```

```python
# 5. Count character occurrences
def exercise_1_5(s, char):
    # Return the number of times 'char' appears in string 's'
    return s.count(char)

# Test the function: count 'l' in 'hello'
print(f"5. Count 'l' in 'hello': {exercise_1_5('hello', 'l')}")
```

```python
# 6. Check if two strings are anagrams
def exercise_1_6(s1, s2):
    # Convert both strings to lowercase and sort their characters
    # If the sorted characters are the same, they are anagrams
    return sorted(s1.lower()) == sorted(s2.lower())

# Test the function with 'listen' and 'silent'
print(f"6. 'listen' and 'silent' are anagrams: {exercise_1_6('listen', 'silent')}")
```

```python
# 7. Remove all vowels
def exercise_1_7(s):
    # Define a string with all vowels (lowercase and uppercase)
    vowels = "aeiouAEIOU"
    # Create an empty string to store the result
    result = ""
    # Loop through each character in the string
    for char in s:
        # If the character is not a vowel, add it to the result
        if char not in vowels:
            result += char
    # Return the string without vowels
    return result

# Test the function: remove vowels from 'hello world'
print(f"7. Remove vowels from 'hello world': {exercise_1_7('hello world')}")
```

```python
# 8. Check if palindrome (case-insensitive)
def exercise_1_8(s):
    # Convert the string to lowercase
    s = s.lower()
    # Check if the string is the same forwards and backwards
    return s == s[::-1]

# Test the function: 'Racecar' is a palindrome
print(f"8. 'Racecar' is palindrome: {exercise_1_8('Racecar')}")
```

```python
# 9. First non-repeating character
def exercise_1_9(s):
    # Loop through each character in the string
    for char in s:
        # If the character appears only once, return it
        if s.count(char) == 1:
            return char
    # If all characters repeat, return None
    return None

# Test the function: first non-repeating in 'aabbcde'
print(f"9. First non-repeating in 'aabbcde': {exercise_1_9('aabbcde')}")
```

```python
# 10. Capitalize first letter of each word
def exercise_1_10(s):
    # Use the title() method to capitalize the first letter of each word
    return s.title()

# Test the function: capitalize 'hello world'
print(f"10. Capitalize 'hello world': {exercise_1_10('hello world')}")
```

```python
# 11. String compression
def exercise_1_11(s):
    # If the string is empty, return an empty string
    if not s:
        return ""
    # Create an empty result string
    result = ""
    # Start counting from the first character
    count = 1
    prev = s[0]
    # Loop through the string starting from the second character
    for i in range(1, len(s)):
        # If the character is the same as the previous, increase the count
        if s[i] == prev:
            count += 1
        else:
            # Add the previous character and its count to the result
            result += prev + str(count)
            # Update the previous character and reset count
            prev = s[i]
            count = 1
    # Add the last character and its count
    result += prev + str(count)
    # Return the compressed string
    return result

# Test the function: compress 'aaabb'
print(f"11. Compress 'aaabb': {exercise_1_11('aaabb')}")
```

```python
# 12. Longest substring without repeating characters
def exercise_1_12(s):
    # Create a set to store unique characters
    char_set = set()
    # Initialize pointers for the sliding window and max length
    left = 0
    max_length = 0
    # Loop through the string with the right pointer
    for right in range(len(s)):
        # If the character is already in the set, remove from the left
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1
        # Add the new character to the set
        char_set.add(s[right])
        # Update the max length if needed
        max_length = max(max_length, right - left + 1)
    # Return the length of the longest substring
    return max_length

# Test the function: longest substring in 'abcabcbb'
print(f"12. Longest substring in 'abcabcbb': {exercise_1_12('abcabcbb')}")
```

```python
# 13. Check if string can be rearranged into palindrome
def exercise_1_13(s):
    # Create a dictionary to count each character
    char_count = {}
    for char in s:
        char_count[char] = char_count.get(char, 0) + 1
    # Count how many characters have an odd count
    odd_count = 0
    for count in char_count.values():
        if count % 2 == 1:
            odd_count += 1
    # If at most one character has an odd count, it can be a palindrome
    return odd_count <= 1

# Test the function: 'aab' can form palindrome
print(f"13. 'aab' can form palindrome: {exercise_1_13('aab')}")
```

```python
# 14. Remove duplicates preserving order
def exercise_1_14(s):
    # Create a set to keep track of seen characters
    seen = set()
    # Create an empty string for the result
    result = ""
    # Loop through each character in the string
    for char in s:
        # If the character is not in seen, add it to result and seen
        if char not in seen:
            seen.add(char)
            result += char
    # Return the string without duplicates
    return result

# Test the function: remove duplicates from 'hello'
print(f"14. Remove duplicates from 'hello': {exercise_1_14('hello')}")
```

```python
# 15. Longest common prefix
def exercise_1_15(strings):
    # If the list is empty, return empty string
    if not strings:
        return ""
    # Start with the first string as the prefix
    prefix = strings[0]
    # Loop through the rest of the strings
    for s in strings[1:]:
        # Shorten the prefix until it matches the start of the string
        while not s.startswith(prefix):
            prefix = prefix[:-1]
            # If prefix is empty, there is no common prefix
            if not prefix:
                return ""
    # Return the longest common prefix
    return prefix

# Test the function: common prefix of ['flower', 'flow', 'flight']
print(f"15. Common prefix of ['flower', 'flow', 'flight']: {exercise_1_15(['flower', 'flow', 'flight'])}")
```

```python
# 1. Dictionary of squares
def exercise_2_1():
    # Create a dictionary with numbers from 1 to 10 as keys and their squares as values
    return {i: i**2 for i in range(1, 11)}

# Print the dictionary of squares
print(f"1. Squares dict: {exercise_2_1()}")
```

```python
# 2. Merge two dictionaries
def exercise_2_2(d1, d2):
    # Copy the first dictionary to a new one
    result = d1.copy()
    # Update the new dictionary with the second dictionary
    result.update(d2)
    # Return the merged dictionary
    return result

# Print the result of merging two dictionaries
print(f"2. Merged dicts: {exercise_2_2({'a': 1}, {'b': 2})}")
```

```python
# 3. Count word occurrences
def exercise_2_3(sentence):
    # Convert the sentence to lowercase and split into words
    words = sentence.lower().split()
    # Create an empty dictionary to store word counts
    word_count = {}
    # Loop through each word in the list
    for word in words:
        # Increase the count for each word
        word_count[word] = word_count.get(word, 0) + 1
    # Return the dictionary with word counts
    return word_count

# Print the word count for the sentence 'hello world hello'
print(f"3. Word count: {exercise_2_3('hello world hello')}")
```

```python
# 4. Dictionary from two lists
def exercise_2_4(keys, values):
    # Use zip to pair keys and values, then convert to dictionary
    return dict(zip(keys, values))

# Print the dictionary created from two lists
print(f"4. Dict from lists: {exercise_2_4(['a', 'b'], [1, 2])}")
```

```python
# 5. Check if key exists
def exercise_2_5(d, key):
    # Return True if the key is in the dictionary, False otherwise
    return key in d

# Print if key 'a' exists in the dictionary {'a': 1}
print(f"5. Key 'a' exists in {{'a': 1}}: {exercise_2_5({'a': 1}, 'a')}")
```

```python
# 6. Invert dictionary
def exercise_2_6(d):
    # Swap keys and values using a dictionary comprehension
    return {v: k for k, v in d.items()}

# Print the inverted dictionary
print(f"6. Inverted dict: {exercise_2_6({'a': 1, 'b': 2})}")
```

```python
# 7. Key with highest value
def exercise_2_7(d):
    # Use max() with key argument to find the key with the highest value
    return max(d, key=d.get)

# Print the key with the max value in the dictionary
print(f"7. Key with max value in {{'a': 1, 'b': 5}}: {exercise_2_7({'a': 1, 'b': 5})}")
```

```python
# 8. Sort dictionary by values
def exercise_2_8(d):
    # Sort the dictionary items by value and convert back to dictionary
    return dict(sorted(d.items(), key=lambda x: x[1]))

# Print the sorted dictionary
print(f"8. Sorted dict: {exercise_2_8({'a': 3, 'b': 1, 'c': 2})}")
```

```python
# 9. Remove duplicates from list
def exercise_2_9(lst):
    # Create a set to keep track of seen items
    seen = set()
    # Create an empty list for the result
    result = []
    # Loop through each item in the list
    for item in lst:
        # If the item is not in seen, add it to result and seen
        if item not in seen:
            seen.add(item)
            result.append(item)
    # Return the list without duplicates
    return result

# Print the list after removing duplicates
print(f"9. Remove duplicates: {exercise_2_9([1, 2, 2, 3, 1])}")
```

```python
# 10. Create set of unique elements
def exercise_2_10(lst):
    # Convert the list to a set to get unique elements
    return set(lst)

# Print the set of unique elements
print(f"10. Unique set: {exercise_2_10([1, 2, 2, 3])}")
```

```python
# 11. Intersection of two sets
def exercise_2_11(s1, s2):
    # Return the intersection of two sets
    return s1.intersection(s2)

# Print the intersection of two sets
print(f"11. Intersection: {exercise_2_11({1, 2, 3}, {2, 3, 4})}")
```

```python
# 12. Check if subset
def exercise_2_12(s1, s2):
    # Return True if s1 is a subset of s2, False otherwise
    return s1.issubset(s2)

# Print if {1, 2} is a subset of {1, 2, 3}
print(f"12. {1, 2} is subset of {1, 2, 3}: {exercise_2_12({1, 2}, {1, 2, 3})}")
```

```python
# 13. Union of two dictionaries
def exercise_2_13(d1, d2):
    # Copy the first dictionary and update with the second
    result = d1.copy()
    result.update(d2)
    # Return the union of both dictionaries
    return result

# Print the union of two dictionaries
print(f"13. Union of dicts: {exercise_2_13({'a': 1}, {'b': 2})}")
```

```python
# 15. Most frequent element
def exercise_2_15(lst):
    # Create a dictionary to count each item
    counts = {}
    for item in lst:
        # Increase the count for each item
        counts[item] = counts.get(item, 0) + 1
    # Return the item with the highest count
    return max(counts, key=counts.get)

# Print the most frequent element in the list
print(f"15. Most frequent in [1, 2, 2, 3]: {exercise_2_15([1, 2, 2, 3])}")
```

```python
# 1. Read and print file contents
def exercise_3_1(filename):
    # Open the file in read mode
    with open(filename, 'r') as f:
        # Read the whole file and print its contents
        print(f.read())

# Create a sample text file for testing
# This file will be used for all Topic 3 exercises
sample_text = """Hello world\nThis is a test file\nPython is fun\nLongestwordhere\n"""
with open('sample_test.txt', 'w') as f:
    f.write(sample_text)
# Call the function to show the file contents using the sample file
exercise_3_1('sample_test.txt')
# The code above creates a file and prints its contents for testing
```

```python
# 2. Count lines in file
def exercise_3_2(filename):
    # Open the file in read mode
    with open(filename, 'r') as f:
        # Read all lines and return the number of lines
        return len(f.readlines())

# Use the sample file created in exercise_3_1 for testing
# This will print the number of lines in the sample file created above
print(f"2. Number of lines: {exercise_3_2('sample_test.txt')}")
# The code above prints the number of lines in the sample file
```

```python
# 3. Count words in file
def exercise_3_3(filename):
    # Open the file in read mode
    with open(filename, 'r') as f:
        # Read the whole file content
        content = f.read()
        # Split the content into words and return the number of words
        return len(content.split())

# Use the sample file created in exercise_3_1 for testing
# This will print the number of words in the sample file created above
print(f"3. Number of words: {exercise_3_3('sample_test.txt')}")
# The code above prints the number of words in the sample file
```

```python
# 4. Count characters in file
def exercise_3_4(filename):
    # Open the file in read mode
    with open(filename, 'r') as f:
        # Read the whole file and return the number of characters
        return len(f.read())

# Use the sample file created in exercise_3_1 for testing
# This will print the number of characters in the sample file created above
print(f"4. Number of characters: {exercise_3_4('sample_test.txt')}")
# The code above prints the number of characters in the sample file
```

```python
# 5. Find longest word
def exercise_3_5(filename):
    # Open the file in read mode
    with open(filename, 'r') as f:
        # Read the file, split into words, and return the longest word
        words = f.read().split()
        return max(words, key=len)

# Use the sample file created in exercise_3_1 for testing
# This will print the longest word in the sample file created above
print(f"5. Longest word: {exercise_3_5('sample_test.txt')}")
# The code above prints the longest word in the sample file
```

```python
# 1. Create DataFrame from dictionary
def exercise_4_1():
    # Create a dictionary with names and ages
    data = {'Name': ['Alice', 'Bob', 'Charlie'], 'Age': [25, 30, 35]}
    # Create a DataFrame from the dictionary
    df = pd.DataFrame(data)
    # Return the DataFrame
    return df

# Print the DataFrame created from the dictionary
print("1. DataFrame from dict:")
print(exercise_4_1())
```

```python
# 2. Add new column
def exercise_4_2(df):
    # Add a new column 'Age_Plus_10' by adding 10 to the 'Age' column
    df['Age_Plus_10'] = df['Age'] + 10
    # Return the updated DataFrame
    return df

# Create the DataFrame using exercise_4_1
# This DataFrame will be used for all Topic 4 exercises
df = exercise_4_1()
# Add the new column and print the result to show the change
print("2. DataFrame with new column:")
print(exercise_4_2(df))
# The code above prints the DataFrame after adding the new column
```

```python
# 4. Sort DataFrame
def exercise_4_4(df, column):
    # Sort the DataFrame by the given column in ascending order
    return df.sort_values(by=column)

# Sort the DataFrame by Age and print the result
print("4. Sorted DataFrame by Age:")
print(exercise_4_4(df, 'Age'))
# The code above prints the sorted DataFrame so you can see the result
```
