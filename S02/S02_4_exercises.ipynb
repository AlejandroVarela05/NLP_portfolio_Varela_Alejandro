### Exercise 1.1: Counting Words

Given the sentence from the slides: *"They picnicked by the pool, then lay back on the grass and looked at the stars."*

Write Python code to:
1. Count words excluding punctuation
2. Count words including punctuation as separate tokens
3. Count unique word types (case-insensitive)

````python
# YOUR CODE HERE
import string


sentence = "They picnicked by the pool, then lay back on the grass and looked at the stars."

# 1. Count words excluding punctuation

cleaned_sentence = sentence.translate(str.maketrans('', '', string.punctuation))
word_count = len(cleaned_sentence.split())
print(f"Word count (excluding punctuation): {word_count}")

# 2. Count words including punctuation
word_count_with_punctuation = len(sentence.split())
print(f"Word count (including punctuation): {word_count_with_punctuation}")

# 3. Count unique types (case-insensitive)
unique_types = set(word.lower() for word in cleaned_sentence.split())
print(f"Unique types: {len(unique_types)}")
````

### Exercise 1.2: Handling Disfluencies

The slides show the utterance: *"I do uh main- mainly business data processing"*

Write code to:
1. Count all tokens including disfluencies
2. Remove filled pauses (uh, um) and fragments (words ending with -)
3. Count "clean" words

````python
# YOUR CODE HERE
utterance = "I do uh main- mainly business data processing"

# 1. Count all tokens including disfluencies
tokens = utterance.split()
print(f"Total tokens (including disfluencies): {len(tokens)}")


# 2. Remove filled pauses (uh, um) and fragments (words ending with -)
cleaned_tokens = [token for token in tokens if token not in ["uh", "um"] and not token.endswith("-")]
print(f"Cleaned tokens: {cleaned_tokens}")


# 3. Count "clean" words
clean_word_count = len(cleaned_tokens)
print(f"Clean word count: {clean_word_count}")

````

### Exercise 2.1: Basic Pattern Matching

Using the regex patterns from the slides, write code to:
1. Find all words starting with capital letters
2. Find all digits in a text
3. Find words that are NOT capitalized

````python
# YOUR CODE HERE
import re


text = "Chapter 1: Down the Rabbit Hole. Alice was 7 years old in 1865."

# 1. Find all words starting with capital letters
capitalized_words = re.findall(r'\b[A-Z][a-z]*\b', text)
print(f"Capitalized words: {capitalized_words}")

# 2. Find all digits in a text
digits = re.findall(r'\d+', text)
print(f"Digits: {digits}")

# 3. Find words that are NOT capitalized
non_capitalized_words = re.findall(r'\b[a-z][a-z]*\b', text)
print(f"Non-capitalized words: {non_capitalized_words}")

````

### Exercise 2.2: Kleene Star and Plus

The slides explain Kleene operators (* and +).

Write patterns to match:
1. "baa" followed by zero or more 'a's (baa, baaa, baaaa...)
2. One or more digits
3. Any sequence of characters (wildcard)

````python
# YOUR CODE HERE
test_strings = ["ba", "baa", "baaa", "baaaa", "baaaaa"]

# 1. "baa" followed by zero or more 'a's (baa, baaa, baaaa...)
pattern1 = r'baaa*'
matches1 = [s for s in test_strings if re.fullmatch(pattern1, s)]
print(f"Matches for pattern 1: {matches1}")

# 2. One or more digits
pattern2 = r'\d+'
test_strings2 = ["123", "456", "abc", "789"]
matches2 = [s for s in test_strings2 if re.fullmatch(pattern2, s)]
print(f"Matches for pattern 2: {matches2}")

# 3. Any sequence of characters (wildcard)
pattern3 = r'.*'
test_strings3 = ["hello", "world", "123", "!@#"]
matches3 = [s for s in test_strings3 if re.fullmatch(pattern3, s)]
print(f"Matches for pattern 3: {matches3}")

````

### Exercise 2.3: Anchors and Word Boundaries

Write patterns using anchors (^, $, \b) to:
1. Match lines starting with a capital letter
2. Match lines ending with a period
3. Find the word "the" as a complete word (not in "other" or "there")

````python
# YOUR CODE HERE
lines = [
    "The quick brown fox",
    "jumped over the lazy dog.",
    "other animals watched.",
    "There was nothing else."
]

# 1. Match lines starting with a capital letter
pattern1 = r'^[A-Z].*'
matches1 = [line for line in lines if re.match(pattern1, line)]
print(f"Lines starting with a capital letter: {matches1}")

# 2. Match lines ending with a period
pattern2 = r'.*\.$'
matches2 = [line for line in lines if re.match(pattern2, line)]
print(f"Lines ending with a period: {matches2}")

# 3. Find the word "the" as a complete word (not in "other" or "there")
pattern3 = r'\bthe\b'
matches3 = []
for line in lines:
    matches3.extend(re.findall(pattern3, line))
print(f"Occurrences of 'the' as a complete word: {matches3}")
````

### Exercise 2.4: Substitutions and Capture Groups

The slides show date format conversion. Write code to:
1. Convert US dates (mm/dd/yyyy) to EU format (dd-mm-yyyy)
2. Anonymize email addresses by replacing with [EMAIL]
3. Convert "I'm" contractions to "I am"

````python
# YOUR CODE HERE
text_dates = "Meeting on 10/15/2011 and 03/22/2024"
text_emails = "Contact john@example.com or jane@test.org"
text_contractions = "I'm happy and I'm excited"

# 1. Convert US dates (mm/dd/yyyy) to EU format (dd-mm-yyyy)
def convert_date(match):
    month, day, year = match.groups()
    return f"{day}-{month}-{year}"
date_pattern = r'(\d{2})/(\d{2})/(\d{4})'
converted_dates = re.sub(date_pattern, convert_date, text_dates)
print(f"Converted dates: {converted_dates}")
# 2. Anonymize email addresses by replacing with [EMAIL]
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
anonymized_emails = re.sub(email_pattern, "[EMAIL]", text_emails)
print(f"Anonymized emails: {anonymized_emails}")
# 3. Convert "I'm" contractions to "I am"
contraction_pattern = r"I'm"
expanded_contractions = re.sub(contraction_pattern, "I am", text_contractions)
print(f"Expanded contractions: {expanded_contractions}")
````
