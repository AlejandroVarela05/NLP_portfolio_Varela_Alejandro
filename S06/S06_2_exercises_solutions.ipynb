### Exercise 1 (Easy)

```python
TP, FP, FN, TN = 40, 5, 10, 45

accuracy = (TP + TN) / (TP + TN + FP + FN)
precision = TP / (TP + FP)
recall = TP / (TP + FN)
f1 = 2 * precision * recall / (precision + recall)

print(f"Accuracy: {accuracy:.2f}")
print(f"Precision: {precision:.2f}")
print(f"Recall: {recall:.2f}")
print(f"F1: {f1:.2f}")
```

### Exercise 2 (Easy)

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

y_true = [1, 1, 1, 0, 0, 0, 1, 0, 1, 0]
y_pred = [1, 0, 1, 0, 1, 0, 1, 0, 0, 0]

print(f"Accuracy: {accuracy_score(y_true, y_pred):.2f}")
print(f"Precision: {precision_score(y_true, y_pred):.2f}")
print(f"Recall: {recall_score(y_true, y_pred):.2f}")
print(f"F1: {f1_score(y_true, y_pred):.2f}")
print(f"Confusion Matrix:\n{confusion_matrix(y_true, y_pred)}")
```

### Exercise 3 (Medium)

```python
from sklearn.metrics import classification_report

y_true = [0, 0, 1, 1, 2, 2, 0, 1, 2, 0, 1, 2]
y_pred = [0, 1, 1, 1, 2, 1, 0, 0, 2, 0, 1, 2]

print(classification_report(y_true, y_pred, target_names=['Negative', 'Neutral', 'Positive']))
```

### Exercise 4 (Medium)

```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

y_true = [0, 0, 1, 1, 0, 1, 0, 1, 1, 0]
y_scores = [0.1, 0.4, 0.35, 0.8, 0.2, 0.9, 0.3, 0.6, 0.7, 0.25]

fpr, tpr, thresholds = roc_curve(y_true, y_scores)
roc_auc = auc(fpr, tpr)

plt.figure()
plt.plot(fpr, tpr, label=f'ROC curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
plt.show()
print(f"AUC: {roc_auc:.2f}")
```

### Exercise 5 (Hard)

```python
from collections import Counter

def macro_precision(y_true, y_pred, num_classes):
    precisions = []
    for c in range(num_classes):
        tp = sum(1 for t, p in zip(y_true, y_pred) if t == c and p == c)
        fp = sum(1 for t, p in zip(y_true, y_pred) if t != c and p == c)
        if tp + fp > 0:
            precisions.append(tp / (tp + fp))
    return sum(precisions) / len(precisions)

def micro_precision(y_true, y_pred, num_classes):
    tp_total = sum(1 for t, p in zip(y_true, y_pred) if t == p)
    # For micro precision = micro recall = accuracy in multiclass
    return tp_total / len(y_true)

y_true = [0, 0, 1, 1, 2, 2, 0, 1, 2, 0, 1, 2]
y_pred = [0, 1, 1, 1, 2, 1, 0, 0, 2, 0, 1, 2]

print(f"Macro Precision: {macro_precision(y_true, y_pred, 3):.3f}")
print(f"Micro Precision: {micro_precision(y_true, y_pred, 3):.3f}")
```
