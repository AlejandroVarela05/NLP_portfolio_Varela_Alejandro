## Exercise 1: Text Generation with Ollama (Easy)

Ollama is a lightweight tool for running LLMs locally. It manages model downloads, quantization, and provides a simple CLI and REST API. Ollama runs models as a local server, exposing an OpenAI-compatible API on `http://localhost:11434`.

Before starting, make sure you have Ollama installed and running on your machine, with a small model pulled (e.g., `gemma3:1b` or `qwen2.5:1.5b`).

**Task:**
1. Use the `requests` library to send a POST request to Ollama's `/api/generate` endpoint
2. Send the prompt: `"Explain what sentiment analysis is in NLP in 2-3 sentences."`
3. Use a small model (e.g., `gemma3:1b`)
4. Set `stream: False` to get the complete response at once
5. Print the model's response text and the total generation time (available in the `total_duration` field of the response, in nanoseconds — convert to seconds)

```python
import requests
import json

OLLAMA_URL = "http://localhost:11434"

# Send a generation request to Ollama
# Model: "gemma3:1b" (or another small model you have pulled)
# Prompt: "Explain what sentiment analysis is in NLP in 2-3 sentences."
# Print the response and generation time
```

```python
import requests
import json
import time

OLLAMA_URL = "http://localhost:11434"

prompt = "Explain what sentiment analysis is in NLP in 2-3 sentences."

payload = {
    "model": "gemma3:1b",   # o "tinyllama" si lo tienes
    "prompt": prompt,
    "stream": False
}

try:
    start_time = time.time()
    response = requests.post(f"{OLLAMA_URL}/api/generate", json=payload, timeout=60)
    elapsed = time.time() - start_time

    if response.status_code == 200:
        data = response.json()
        print("Ollama response:")
        print(data["response"])
        # total_duration está en nanosegundos
        total_ns = data.get("total_duration", 0)
        print(f"Generation time (from API): {total_ns / 1e9:.2f} seconds")
        print(f"Measured round-trip: {elapsed:.2f} seconds")
    else:
        print(f"Error {response.status_code}: {response.text}")
except Exception as e:
    print(f"Ollama not available: {e}")
```

## Exercise 2: Chat Completion with LM Studio (Easy)

LM Studio is a desktop application that lets you download, manage, and run LLMs locally with a graphical interface. Like Ollama, it exposes a local API server — but LM Studio uses the OpenAI-compatible chat completions format at `http://localhost:1234/v1`.

Before starting, make sure you have LM Studio installed, a small model loaded (e.g., a quantized Qwen, Gemma or Phi model), and the local server started.

**Task:**
1. Use the `openai` Python library to connect to LM Studio's local server
2. Set the `base_url` to `http://localhost:1234/v1` and use `api_key="lm-studio"` (any non-empty string works)
3. Send a chat completion request with a system message: `"You are an NLP expert."` and a user message: `"What are the main challenges in sentiment analysis?"`
4. Use the model identifier shown in LM Studio (you can also use `model="local-model"` as a generic identifier)
5. Print the response content and the token usage statistics (prompt tokens, completion tokens, total tokens)

```python
from openai import OpenAI

# Connect to LM Studio's local server
# Send a chat completion request about sentiment analysis challenges
# Print the response and token usage
```

```python
from openai import OpenAI
import time

# Conectar a LM Studio
client = OpenAI(base_url="http://localhost:1234/v1", api_key="lm-studio")

messages = [
    {"role": "system", "content": "You are an NLP expert."},
    {"role": "user", "content": "What are the main challenges in sentiment analysis?"}
]

start_time = time.time()
response = client.chat.completions.create(
    model="local-model",
    messages=messages,
    temperature=0.7
)
end_time = time.time()

print("LM Studio response:")
print(response.choices[0].message.content)
print(f"\nResponse time: {end_time - start_time:.2f} seconds")

# Token usage (si el servidor lo proporciona)
if hasattr(response, 'usage') and response.usage:
    print(f"Token usage: prompt={response.usage.prompt_tokens}, "
          f"completion={response.usage.completion_tokens}, "
          f"total={response.usage.total_tokens}")
else:
    print("Token usage not available from this server.")
```

```python
import time
import requests
from openai import OpenAI
from transformers import pipeline

common_prompt = (
    "Classify the sentiment of the following text as Positive, Negative, or Neutral. "
    "Only respond with the sentiment label.\n\n"
    "Text: 'The movie was absolutely wonderful, I loved every minute of it'\n\n"
    "Sentiment:"
)

results = {}

# ------------------ Ollama ------------------
print("Querying Ollama...")
try:
    start = time.time()
    ollama_resp = requests.post(
        "http://localhost:11434/api/generate",
        json={"model": "gemma3:1b", "prompt": common_prompt, "stream": False},
        timeout=30
    )
    ollama_time = time.time() - start
    if ollama_resp.status_code == 200:
        raw = ollama_resp.json()["response"].strip()
        # Extraer etiqueta
        if "positive" in raw.lower():
            ollama_label = "Positive"
        elif "negative" in raw.lower():
            ollama_label = "Negative"
        elif "neutral" in raw.lower():
            ollama_label = "Neutral"
        else:
            ollama_label = raw[:20]
    else:
        ollama_label = f"Error {ollama_resp.status_code}"
    results["Ollama"] = {"response": ollama_label, "time": ollama_time}
except Exception as e:
    results["Ollama"] = {"response": f"Not available: {e}", "time": 0}

# ------------------ LM Studio ------------------
print("Querying LM Studio...")
client_lm = OpenAI(base_url="http://localhost:1234/v1", api_key="lm-studio")
start = time.time()
lm_resp = client_lm.chat.completions.create(
    model="local-model",
    messages=[{"role": "user", "content": common_prompt}],
    temperature=0.0,
    max_tokens=10
)
lm_time = time.time() - start
raw_lm = lm_resp.choices[0].message.content.strip()
if "positive" in raw_lm.lower():
    lm_label = "Positive"
elif "negative" in raw_lm.lower():
    lm_label = "Negative"
elif "neutral" in raw_lm.lower():
    lm_label = "Neutral"
else:
    lm_label = raw_lm[:20]
results["LM Studio"] = {"response": lm_label, "time": lm_time}

# ------------------ HuggingFace ------------------
print("Querying HuggingFace...")
hf_pipeline = pipeline("text-generation", model="TinyLlama/TinyLlama-1.1B-Chat-v1.0", device="cpu")
start = time.time()
hf_out = hf_pipeline(common_prompt, max_new_tokens=10, do_sample=False, temperature=0.0)
hf_time = time.time() - start
raw_hf = hf_out[0]["generated_text"].replace(common_prompt, "").strip()
if "positive" in raw_hf.lower():
    hf_label = "Positive"
elif "negative" in raw_hf.lower():
    hf_label = "Negative"
elif "neutral" in raw_hf.lower():
    hf_label = "Neutral"
else:
    hf_label = raw_hf[:20]
results["HuggingFace"] = {"response": hf_label, "time": hf_time}

# ------------------ Tabla comparativa ------------------
print("\n" + "="*60)
print("Comparison Table")
print("="*60)
for tool, data in results.items():
    print(f"\n{tool}:")
    print(f"  Response: {data['response']}")
    print(f"  Time (s): {data['time']:.2f}")

print("\n" + "="*60)
print("Observations:")
print("- Ollama: Did not work on my machine (memory/GPU issues). The code is correct but requires a working Ollama server.")
print("- LM Studio: Very easy to use, fast, and gives clean responses. The GUI is intuitive.")
print("- HuggingFace: No external server needed, but slower and sometimes produces extra text.")
print("\nEasiest to use: LM Studio (just start the server and use OpenAI client).")
print("Best response quality: LM Studio (consistent label output).")
print("Trade-offs: HuggingFace is fully Python-integrated but heavier; LM Studio requires a separate app but is more robust.")
```

```python
import time
import requests
import numpy as np
from openai import OpenAI
from transformers import pipeline
from sklearn.metrics import classification_report, confusion_matrix

# ------------------ Test dataset (12 ejemplos) ------------------
test_data = [
    ("I absolutely love this product! It's fantastic.", "Positive"),
    ("This is the worst experience I've ever had.", "Negative"),
    ("The movie was okay, nothing special.", "Neutral"),
    ("She performed brilliantly on stage.", "Positive"),
    ("I hate waiting in long queues.", "Negative"),
    ("The book was mediocre, not good but not terrible.", "Neutral"),
    ("What a wonderful day to be alive!", "Positive"),
    ("The service was slow and the food cold.", "Negative"),
    ("It's neither good nor bad, just average.", "Neutral"),
    ("Excellent work, keep it up!", "Positive"),
    ("This is terrible, I want a refund.", "Negative"),
    ("The presentation was fine.", "Neutral"),
]

# Función auxiliar para extraer etiqueta de cualquier texto
def extract_label(text):
    text_lower = text.lower()
    if "positive" in text_lower:
        return "Positive"
    elif "negative" in text_lower:
        return "Negative"
    elif "neutral" in text_lower:
        return "Neutral"
    else:
        return "Unknown"   # por si el modelo no responde bien

# ------------------ Predictor para Ollama (si funciona) ------------------
def predict_ollama(text):
    prompt = f"Classify the sentiment of the following text as Positive, Negative, or Neutral. Only respond with the sentiment label.\n\nText: '{text}'\n\nSentiment:"
    try:
        resp = requests.post(
            "http://localhost:11434/api/generate",
            json={"model": "gemma3:1b", "prompt": prompt, "stream": False, "options": {"temperature": 0.0}},
            timeout=30
        )
        if resp.status_code == 200:
            raw = resp.json()["response"].strip()
            return extract_label(raw)
        else:
            return "Unknown"
    except:
        return "Unknown"

# ------------------ Predictor para LM Studio ------------------
client_lm = OpenAI(base_url="http://localhost:1234/v1", api_key="lm-studio")
def predict_lmstudio(text):
    prompt = f"Classify the sentiment of the following text as Positive, Negative, or Neutral. Only respond with the sentiment label.\n\nText: '{text}'\n\nSentiment:"
    try:
        resp = client_lm.chat.completions.create(
            model="local-model",
            messages=[{"role": "user", "content": prompt}],
            temperature=0.0,
            max_tokens=10
        )
        raw = resp.choices[0].message.content.strip()
        return extract_label(raw)
    except:
        return "Unknown"

# ------------------ Predictor para HuggingFace ------------------
hf_pipeline = pipeline("text-generation", model="TinyLlama/TinyLlama-1.1B-Chat-v1.0", device="cpu")
def predict_huggingface(text):
    prompt = f"Classify the sentiment of the following text as Positive, Negative, or Neutral. Only respond with the sentiment label.\n\nText: '{text}'\n\nSentiment:"
    try:
        out = hf_pipeline(prompt, max_new_tokens=10, do_sample=False, temperature=0.0)
        generated = out[0]["generated_text"]
        answer = generated[len(prompt):].strip()
        return extract_label(answer)
    except:
        return "Unknown"

# ------------------ Evaluación ------------------
tools = {
    "Ollama": predict_ollama,
    "LM Studio": predict_lmstudio,
    "HuggingFace": predict_huggingface
}

true_labels = [label for _, label in test_data]
texts = [text for text, _ in test_data]

for tool_name, pred_func in tools.items():
    print(f"\n--- Evaluating {tool_name} ---")
    predictions = []
    times = []
    for text in texts:
        start = time.time()
        pred = pred_func(text)
        elapsed = time.time() - start
        predictions.append(pred)
        times.append(elapsed)
        print(f"Text: {text[:50]}... -> {pred}")
    
    print(f"\nAverage inference time per sample: {np.mean(times):.2f} seconds")
    print("\nClassification Report:")
    # Filtrar solo las clases presentes (Positive, Negative, Neutral)
    print(classification_report(true_labels, predictions, labels=["Positive", "Negative", "Neutral"]))
    print("Confusion Matrix (rows=true, cols=pred):")
    cm = confusion_matrix(true_labels, predictions, labels=["Positive", "Negative", "Neutral"])
    print(cm)
    print("-" * 60)

print("\n" + "="*60)
print("FINAL ANALYSIS")
print("="*60)
print("""
Which tool performed best?
- LM Studio gave the most consistent and accurate labels, with very few parsing errors.
- HuggingFace also performed well but sometimes produced extra words.
- Ollama did not work on my machine (persistent memory/GPU errors).

Failure modes observed:
- HuggingFace occasionally outputs full sentences instead of just the label; extract_label() handles this.
- LM Studio sometimes returned 'Neutral' for obviously positive texts (rare).
- Ollama (when it worked briefly) was slow and often crashed.

How parsing was handled:
- A simple keyword search for 'positive', 'negative', 'neutral' in the lowercased response.
- If none found, returns 'Unknown' (which counts as an error in metrics).

Recommendation for local NLP tasks:
- Use LM Studio if you want a reliable, fast, GUI-based tool with an OpenAI-compatible API.
- Use HuggingFace if you need deep integration with Python and don't mind slower inference.
- Avoid Ollama on Windows with limited GPU memory; it requires careful configuration.
""")
```
